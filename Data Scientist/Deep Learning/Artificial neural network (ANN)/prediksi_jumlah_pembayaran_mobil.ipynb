```python
import pandas as pd
import tensorflow as tf
import numpy as np
```

```python
# Getting the dataset
df = pd.read_csv(
    "car_purchasing.csv", encoding='latin1')
```

```python
X = df.drop(["customer name", "customer e-mail", "country", "car purchase amount" ,"gender"], axis=1).values
y = df["car purchase amount"].values
```

```python
# Splitting the dataset
X_train, y_train = X[:350], y[:350]
X_valid, y_valid = X[350:450], y[350:450]
X_test, y_test = X[450:], y[450:]
```

```python
# Set random seed for reproducibility
tf.random.set_seed(42)

# Normalization layer
norm_layer = tf.keras.layers.Normalization(input_shape=X_train.shape[1:])
norm_layer.adapt(X_train)  # Adapt the normalization layer to the training data

# Model definition
model = tf.keras.Sequential([
    norm_layer,  # Normalization layer
    tf.keras.layers.Dense(50, activation="relu"),
    tf.keras.layers.Dense(50, activation="relu"),
    tf.keras.layers.Dense(50, activation="relu"),
    tf.keras.layers.Dense(1)  # Output layer
])
```

```python
# Compile the model
model.compile(optimizer="nadam", loss="mse", metrics=["mae"])  # Mean Absolute Error for metrics

# Train the model
history = model.fit(
    X_train, y_train,
    epochs=200,
    validation_data=(X_valid, y_valid)
)
```

```python
# Evaluate the model on the test set
mse_test = model.evaluate(X_test, y_test, verbose=0)[0]  # MSE
rmse_test = np.sqrt(mse_test)  # RMSE

y_pred = model.predict(X_test)
for i in range(len(X_test)):
    print(y_test[i] ,"vs" , y_pred[i])

from sklearn.metrics import r2_score
r2 = r2_score(y_test,y_pred)
print('score:', r2)
```
